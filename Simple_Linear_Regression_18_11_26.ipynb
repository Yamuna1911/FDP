```python

from sklearn.linear_model import LinearRegression
X=[[1],[2],[3]]
Y=[[3],[6],[9]]
model=LinearRegression()
model.fit(X,Y)
print(model.predict([[6]]))
```

```python
#To calculate m and c
m=model.coef_[0]
c=model.intercept_
print('Slope:',m)
print('Intercept:',c)
print('Predict',model.predict([[6]]))
```
