```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load dataset
df = pd.read_csv('/content/drive/MyDrive/DengueDB.csv')
```

```python
print(df.info())
```

```python
print(df.describe())
```

```python
# 1. Box Plot
# 7. Box Plot
plt.figure(figsize=(6,3))

plt.boxplot(df['treatment_cost_inr'])

plt.title('Box Plot of Treatment Cost')
plt.ylabel('Treatment Cost (INR)')

plt.show()
```

```python
# 2. Scatter Plot
plt.figure(figsize=(8,5))

plt.scatter(df['age'], df['treatment_cost_inr'])

plt.title('Age vs Treatment Cost')
plt.xlabel('Age')
plt.ylabel('Treatment Cost (INR)')

plt.show()

```

```python
# 3. Histogram
plt.figure(figsize=(8,5))

plt.hist(df['treatment_cost_inr'], bins=20, edgecolor='black')

plt.title('Distribution of Treatment Cost')
plt.xlabel('Treatment Cost (INR)')
plt.ylabel('Number of Patients')

plt.show()
```

```python
# 4. Pie Chart
disease_count = df['disease_name'].value_counts()

plt.figure(figsize=(7,7))

plt.pie(
    disease_count,
    labels=disease_count.index,
    autopct='%1.1f%%',
    startangle=90
)

plt.title('Distribution of Patients by Disease')
plt.show()
```

```python
# 5. Bar Chart
plt.figure(figsize=(10,5))

plt.bar(disease_count.index, disease_count.values)

plt.title('Number of Patients by Disease')
plt.xlabel('Disease')
plt.ylabel('Number of Patients')

plt.xticks(rotation=45)
plt.show()
```

```python
# 6. Line Plot
year_cost = df.groupby('year')['treatment_cost_inr'].mean()

plt.figure(figsize=(10,5))

plt.plot(
    year_cost.index,
    year_cost.values,
    marker='o'
)

plt.title('Average Treatment Cost by Year')
plt.xlabel('Year')
plt.ylabel('Average Treatment Cost (INR)')

plt.grid()
plt.show()
```
